```python
from pathlib import Path
import json

IMAGE_FOLDER = Path("product-images")
OUTPUT_FILE = Path("products.js")

ALLOWED_EXTENSIONS = {".png", ".jpg", ".jpeg", ".webp", ".gif"}

products = []

for image_path in IMAGE_FOLDER.iterdir():
    if image_path.is_file() and image_path.suffix.lower() in ALLOWED_EXTENSIONS:
        products.append({
            "name": image_path.stem,
            "image": f"product-images/{image_path.name}"
        })

products = sorted(products, key=lambda item: item["name"].lower())

js_content = "window.PRODUCTS = "
js_content += json.dumps(products, ensure_ascii=False, indent=2)
js_content += ";\n"

OUTPUT_FILE.write_text(js_content, encoding="utf-8")

print(f"Done. {len(products)} product image(s) found.")
print("Generated file: products.js")
```
